# RECALL PRACTICE


```python
relevant = ["A", "B", "C"]
ranked = ["X", "A", "Y", "B", "Z"]
k = 3

intersection =set(relevant) & set(ranked[:k])
ans=len(intersection)/len(set(relevant))

ans
```

# RECIPROCAL RANK

```python
ranked = ["X", "Y", "A", "B", "C"]
relevant = ["A", "B", "C"]
k = 5
rel=set(relevant)

def reciprocal_rank(ranked, relevant, k=None):
    for i,c in enumerate(ranked[:k],start=1):
        if c in rel:
            return 1/i
    return 0

reciprocal_rank(ranked, relevant, k)
```
